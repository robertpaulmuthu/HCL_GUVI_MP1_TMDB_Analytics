## Data Cleaning

### Change data type in Movies data set

```python
df_movies['release_date'] = pd.to_datetime( df_movies['release_date'])
```

```python
df_movies.info()
```

```python
df_movies['release_date'].isna().sum()
```

###Fix the budget values which are zero

```python
df_movies['budget'].max()
```

```python
df_movies['budget'].min()
```

```python
(df_movies['budget'] == 0).sum()
```

```python
budget_data = df_movies[df_movies['budget'] > 0]['budget']
```

```python
budget_data.describe()
```

```python
Q1 = budget_data.quantile(0.25)
Q3 = budget_data.quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
clean_budget = budget_data[
    (budget_data >= lower_limit) &
    (budget_data <= upper_limit)
]

median_budget = clean_budget.median()

print("Median Budget:", median_budget)
```

```python
df_movies['budget'] = df_movies['budget'].replace(0, median_budget)
```

```python
(df_movies['budget'] == 0).sum()
```
