##Question 3
List the profit (revenue − budget) and return on investment for every movie that has valid budget and revenue figures.

```python
valid_movies = movies[
    (movies['budget'] > 0) &
    (movies['revenue'] > 0)
].copy()
```

```python
valid_movies['profit'] = (
    valid_movies['revenue'] - valid_movies['budget']
)
```

```python
valid_movies['roi'] = (
    valid_movies['profit'] / valid_movies['budget']
) * 100
```

```python
profit_roi = valid_movies[
    ['movie_id', 'title', 'budget', 'revenue', 'profit', 'roi']
]

profit_roi
```

```python
profit_roi.sort_values(
    by='roi',
    ascending=False
)
```

```python
highest_profit = profit_roi.sort_values(
    by='profit',
    ascending=False
).head(1)

highest_profit
```

```python
profit_count = (profit_roi['profit'] > 0).sum()

print("Total Profitable Movies:", profit_count)
```

```python
loss_count = (profit_roi['profit'] < 0).sum()

print("Total Loss Movies:", loss_count)
```

```python
highest_loss = profit_roi.sort_values(
    by='profit',
    ascending=True
).head(1)

highest_loss
```

####Insight
2370 movies have valid positive budget and revenue values. Profit is revenue minus budget. ROI measures profit relative to the original budget. Also
1. Highest Profit Movie: Avatar

2. Highest Loss Movie: Wake Up Dead Man: A Knives Out Mystery

3. Total Profitable Movies: 2182

4. Total Loss Movies: 194

##Question 6
List all movies whose runtime is a statistical outlier compared to the rest of the dataset.

```python
runtime_data = movies[
    movies['runtime'].notnull() &
    (movies['runtime'] > 0)
]

Q1 = runtime_data['runtime'].quantile(0.25)

Q3 = runtime_data['runtime'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - (1.5 * IQR)

upper_limit = Q3 + (1.5 * IQR)

print("Q1:", Q1)

print("Q3:", Q3)

print("IQR:", IQR)

print("Lower Limit:", lower_limit)

print("Upper Limit:", upper_limit)
```

```python
runtime_outliers = runtime_data[
    (runtime_data['runtime'] < lower_limit) |
    (runtime_data['runtime'] > upper_limit)
]
```

```python
runtime_outliers[
    ['title', 'runtime']
].sort_values(
    by='runtime',
    ascending=False
)
```

```python
print("Total Runtime Outliers:", len(runtime_outliers))
```

####Insight
60 movies are runtime outliers using the IQR method. Values below 58.50 or above 166.50 minutes are flagged. These should be investigated, not automatically deleted.

##Question 7
List each movie's primary genre (the first genre listed) and compare average popularity across primary genres.

```python
movie_genre_data = pd.merge(
    movie_genres,
    genres,
    on='genre_id',
    how='left'
)
```

```python
movie_genre_data = pd.merge(
    movie_genre_data,
    movies[['movie_id', 'title', 'popularity']],
    on='movie_id',
    how='left'
)
```

```python
movie_genre_data['genre_order'] = (
    movie_genre_data.groupby('movie_id').cumcount()
)
```

```python
primary_genre = movie_genre_data[
    movie_genre_data['genre_order'] == 0
].copy()
```

```python
primary_genre[
    ['movie_id', 'title', 'genre_name', 'popularity']
]
```

```python
genre_popularity = (
    primary_genre
    .groupby('genre_name')['popularity']
    .mean()
    .sort_values(ascending=False)
)
```

```python
print(genre_popularity)
```

####Insight
Science Fiction has the highest average popularity among primary genres.

##Question 8
List the top 10 most prolific actors by number of movies, along with their average movie rating across those movies.

```python
cast_movies = pd.merge(
    cast,
    movies[['movie_id', 'vote_average']],
    on='movie_id',
    how='inner'
)
```

```python
actor_analysis = (
    cast_movies
    .groupby('actor_name')
    .agg(
        total_movies=('movie_id', 'nunique'),
        average_rating=('vote_average', 'mean')
    )
    .reset_index()
)
```

```python
actor_analysis = actor_analysis.sort_values(
    by='total_movies',
    ascending=False
)
```

```python
top_10_actors = actor_analysis.head(10)

top_10_actors
```

####Insight
Samuel L. Jackson has appeared in the highest number of movies among the top 10 actors.

##Question 10
List movies whose popularity score and vote count don't follow the general pattern between the two — i.e., unusually high popularity with a low vote count, or vice versa.

```python
popularity_analysis = movies[
    ['movie_id', 'title', 'popularity', 'vote_count']
].copy()
```

```python
popularity_analysis = popularity_analysis.dropna(
    subset=['popularity', 'vote_count']
)
```

```python
median_popularity = popularity_analysis['popularity'].median()

median_votes = popularity_analysis['vote_count'].median()

print("Median Popularity:", median_popularity)

print("Median Vote Count:", median_votes)
```

```python
high_pop_low_votes = popularity_analysis[
    (popularity_analysis['popularity'] > median_popularity) &
    (popularity_analysis['vote_count'] < median_votes)
]
```

```python
low_pop_high_votes = popularity_analysis[
    (popularity_analysis['popularity'] < median_popularity) &
    (popularity_analysis['vote_count'] > median_votes)
]
```

```python
print("High Popularity with Low Vote Count")

high_pop_low_votes
```

```python
print("Low Popularity with High Vote Count")

low_pop_high_votes
```

```python
unusual_movies = pd.concat([
    high_pop_low_votes,
    low_pop_high_votes
])

unusual_movies
```

####Insight
The analysis identified movies with unusual combinations of popularity and vote counts. Some movies had higher popularity despite receiving fewer votes, while others received more votes despite having lower popularity. These results highlight differences in audience engagement and popularity patterns.
